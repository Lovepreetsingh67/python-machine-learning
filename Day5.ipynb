```python
#variable length arguments using args
def calculate_total_expenses(*expenses):
    print(f"Expenses received:{expenses} (type: {type(expenses)})")
    return sum(expenses)

total=calculate_total_expenses(12.5,45.0,100.25,7.8)
print(f"Total:{total}\n")
```

```python
#Example of **kwargs
def print_model_parameters(**hyperparams):
    print(f"Hyperparameters received: {hyperparams} (type:{type(hyperparams)})")
    for key, value in hyperparams.items():
        print(f"  -{key}: {value}")

print_model_parameters(learning_rate=0.001, batch_size=64, dropout=0.3)
```

```python
#global variable
metric_name="Accuracy"

def evaluate():
    #local variable
    score=0.95
    print(f"Inside function: {metric_name} score is {score}")

evaluate()
#testing scope access ouutside
print(metric_name)
#print(score)   #show error
```

```python
#copy of global variable
a=10
def change():
    a=20         #copy of global variable
    print("after changing tha 'a' variable:",a)

change()
print("original : ",a)
```

```python
#increments the global variable
epoch_counter=5   #global variable

def run_epoch():
    global epoch_counter
    epoch_counter +=1  #increments the global variable
    print(f"Inside run_epoch: epoch_counter incremented to {epoch_counter} ")
run_epoch()
print(f"Before: epoch_counter = {epoch_counter}")
```

```python
#recursive function to calculate
def calculate_factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n* calculate_factorial(n-1)

print(calculate_factorial(5))
```

```python
#fibonacci sequence
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

n = int(input("Enter number of Terms: "))
print(fib(n))
for i in range(n+1):
    print(fib(i), end=" ")
```
