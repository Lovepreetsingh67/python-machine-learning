```python
import numpy as np
#1D array
list1=[10,20,40,50]
arr=np.array(list1)
print(arr)
print(type(arr))
print(list1)
```

```python
#2D array
list2=[[1,2,3],
       [4,5,6],
       [7,8,9]]
arr2d=np.array(list2)
print(arr2d)
print(list2)
print(type(arr2d))
```

```python
#Creating an array with float data type
float_arr= np.array([1,2,3], dtype=np.float64)
print(float_arr)
print(float_arr.dtype)
```

```python
#Array Attributes (methods)
my_matrix= np.array(([[1,2,3],[4,5,6]]),dtype=np.int32)

print(my_matrix)
print()
print(my_matrix.ndim)      #its give number of dimensions of an array
print()
print(my_matrix.shape)     #its give shape of an array
print()
print(my_matrix.size)      #its give size of an array
print()
print(my_matrix.dtype)     #its give datatype of an array
print()
print(my_matrix.itemsize)  #its give bytes storage in memory of an array
```
