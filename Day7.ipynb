```python
#assignment 1 temperature cold days
tem_list = [28, 32, 18, 15, 30, 22, 17]

cold_days = []

for temp in tem_list:
    if temp < 20:
        cold_days.append(temp)

print("Cold days:", cold_days)
print("Number of cold days:", len(cold_days))


#hot days
temperatures = [28, 32, 18, 15, 30, 22, 17]

hot_days = []

for temp in temperatures:
    if temp > 30:
        hot_days.append(temp)

print("Hot days:", hot_days)
print("Number of hot days:", len(hot_days))
```
