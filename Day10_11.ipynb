```python
import io
csv_data="""customerid,age,gender,income,purchased
101,25,F,50000.0,no
102,47,M,85000.0,yes
103,,F,62000.0,yes
104,31,M,,no
105,19,F,22000.0,no
106,52,,98000.0, yes
"""
print(csv_data)
```
