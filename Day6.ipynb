```python
#filtering outlier
predictions=[0.15, -99.0,0.82,1.45, -99.0, -0.67,2.1]
sorted_list=sorted(predictions)
print(sorted_list)
outlier=sorted_list[0]
while outlier in predictions:
  predictions.remove(outlier)
  print(predictions)
```

```python
#Set operatiosns
train_classes={"cat","dog","bird","cat"}
print(train_classes)
print(type(train_classes))

val_classes={"dog","horse","sheep"}

all_classes=train_classes.union(val_classes)
print(all_classes)
common_classes=train_classes.intersection(val_classes)
print(common_classes)
```

```python
train_classes={"cat","dog","bird","cat"}
print(train_classes)
print(type(train_classes))

val_classes={"dog","horse","sheep"}

all_classes=train_classes | val_classes
print(f"All classes= {all_classes}")
common_classes=train_classes & val_classes
print(f"Common classes= {common_classes}")
only_train=train_classes-val_classes
print(f"Only train classes= {only_train}")
```
