```python
#calculator
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b

a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))

print("Sum =", add(a, b))
print("Sub =", sub(a, b))
print("Mul =", mul(a, b))
print("Div =", div(a, b))
```

```python
#prime number
from math import sqrt
def is_prime(n):
    if n<=1:
        return False

    for i in range(2, int(sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(11))
#call the function with the value 11
result=is_prime(11)
```
