```python
a=15
b=4
print(f"Division:{a}/{b}= {a/b}")
print(f"Floor Division:{a}//{b}= {a//b}")
print(f"Modulus(Remainder):{a}%{b}= {a%b}")
print(f"Exponentiation:{a}**{b}= {a**b}")
```

```python
#Even or odd number
num=31

if num%2==0:
  print("Even Number")
else:
  print("Odd Number")
```

```python
#Comparison(Relational) operator
x=10
y=20

print(f"X==Y(Equal): {x==y}")
print(f"X!=Y(Not Equal): {x!=y}")
print(f"X>Y(Greater Than): {x>y}")
print(f"X<Y(Less Than): {x<y}")
print(f"X>=Y(Greater Than or Equal): {x>=y}")
print(f"X<=Y(Less Than or Equal): {x<=y}")
```

```python
#Logical operator
is_sunny=True
is_weekend=False

print(f"Is it a sunny day and not a weekend: {is_sunny and not is_weekend}")
print(f"Is it a sunny day or not a weekend: {is_sunny or not is_weekend}")
print(f"Is it a sunny day and a weekend: {is_sunny and is_weekend}")
```

```python
#Assignment operator
num=10
num+=5
print(f"After +=5: {num}")
num-=5
print(f"After -=5: {num}")
num*=5
print(f"After *=5: {num}")
num/=5
print(f"After /=5: {num}")
num//=3
print(f"  After //=3: {num}")
num**=2
print(f"After **=2: {num}")
```

```python
#bitwise operator
a=13
b=11

print(f"Bitwise AND: {a&b}")
print(f"Bitwise OR: {a|b}")
print(f"Bitwise XOR: {a^b}")
print(f"Bitwise NOT: {~a}")
print(f"Bitwise Left Shift: {a<<2}")
print(f"Bitwise Right Shift: {a>>2}")
```

```python
#Membership operator
fruits=["apple","banana","cherry"]

print("banana" in fruits)
print("orange" in fruits)
print("mango" not in fruits)
```

```python
#identity operator
list_a=[1,2,3]
list_b=[1,2,3]
list_c=list_a

print(f"list_a is list_b: {list_a is list_b}")
print(f"list_a is list_c: {list_a is list_c}")
print(f"list_a == list_b: {list_a == list_b}")
print(f"list_a is not list_b: {list_a is not list_b}")
```

```python
num=30

if num>0 & num%2==0:
  print("Positive even number")
elif num>0 & num%2!=0:
  print("Positive odd number")
```

```python
num=30

if num%5==0 | num%2==0:
  print(True)
else:
  print(False)
```

```python
def is_even(num):
  return(num & 1)==0
def is_odd(num):
  return(num &1)==1

print(is_even(4))
print(is_odd(7))
```
