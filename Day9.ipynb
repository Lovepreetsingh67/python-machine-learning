```python
x=np.array([
    [10.0,  200.0],
    [30.0,  100.0],
    [20.0,  300.0]
])
row_means= np.mean(x, axis=0)
print(row_means)
col_means= np.mean(x, axis=1)
print(col_means)
```

```python
flat_arr=np.arange(6)
print("Original 1D array:", flat_arr)

#reshape to 2D matrix of 2 rows and 3 columns
matrix_2d=flat_arr.reshape(2,3)
print("\nReshaped to 2x3 Matrix:\n",matrix_2d)

#flatten it back to 1d
flattened= matrix_2d.flatten()
print("\nFlattened to 1D array:",flattened)
```
